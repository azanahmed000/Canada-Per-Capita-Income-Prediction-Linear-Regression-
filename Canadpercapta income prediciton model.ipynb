```python
plt.title("Year per capita income of Canda(1917-2016)",fontsize=14,fontweight="bold")
plt.scatter(df["year"],df["per capita income (US$)"],marker='.',color='red',label='Canada Income Data')
plt.xlabel("year")
plt.ylabel("income in US($)")
plt.grid(True,alpha=0.7)
plt.legend()
plt.show()
```
